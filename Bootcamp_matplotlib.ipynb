```python
import matplotlib.pyplot as plt
import numpy as np

from matplotlib import cm

plt.style.use('_mpl-gallery')

# Make data
X = np.arange(-5, 5, 0.25)
Y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(X, Y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

# Plot the surface
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot_surface(X, Y, Z, vmin=Z.min() * 2, cmap=cm.Blues)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

plt.style.use('_mpl-gallery')

# Fix: Change "delhi_ncr_aqi_dataset.csv" to "/delhi_ncr_aqi_dataset.csv" for the correct file path
df = pd.read_csv('/delhi_ncr_aqi_dataset.csv')

# Fix: Change "AQI" to "aqi" to match the column name in the DataFrame
grouped = df.groupby("city")["aqi"].mean()

x = np.arange(len(grouped))
y = grouped.values

fig, ax = plt.subplots()

ax.bar(x, y, width=0.8, edgecolor="white", linewidth=0.7)

ax.set_xticks(x)
ax.set_xticklabels(grouped.index, rotation=45)

ax.set_ylabel("Average AQI")
ax.set_title("Average AQI by City (Delhi NCR)")

plt.tight_layout()
plt.show()
```
